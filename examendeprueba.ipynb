```python
### r es una tasa de crecimeitno que podemos cambiar por cualñquiier numero
import numpy as np
import matplotlib.pyplot as plt
r = 3.8
x = np.linspace(0, 1, 200)

plt.plot(x, r * x * (1 - x), label=f'r = {r}')
plt.plot(x, x, '--', label='$x_{n+1} = x_n$')

plt.xlabel('$x_n$')
plt.ylabel('$x_{n+1}$')
plt.legend()
plt.title("crecimiento logistico")
plt.show()
```

```python
##Xn vs n para n e 0,...,1000

r = 3.8
x = [0.2]  # valor inicial x_0

for _ in range(1000):
    x.append(r * x[-1] * (1 - x[-1]))

plt.figure(figsize=(10, 4))
plt.plot(x, linewidth=0.7)
plt.xlabel('n')
plt.ylabel('$x_n$')
plt.title(f'Mapeo logístico: $x_n$ vs $n$ (r = {r})')
plt.show()
```
