# Tasa de crecimiento


El tiempo de duplicación (td) es el tiempo necesario para que una población se multiplique por dos.

Para un crecimiento exponencial:

N(t) = N₀ e^(rt)

Si conocemos el tiempo de duplicación, podemos calcular la tasa de crecimiento:

r = ln(2) / td

donde:
* r = tasa de crecimiento
* td = tiempo de duplicación
* ln(2) = logaritmo natural de 2

Por lo tanto:


```python
td_A = 1
td_B = 2
```

# Calcular la tasa de crecimiento

```python
r_A = np.log(2) / td_A
r_B = np.log(2) / td_B
```

```python
print("Tasa de crecimiento de A:", r_A, "h-1")
print("Tasa de crecimiento de B:", r_B, "h-1")
```

**Para la población A:**

td = 1 hora

rA = ln(2) / 1 = **0.693 h⁻¹**

**Para la población B:**

td = 2 horas

rB = ln(2) / 2 = **0.347 h⁻¹**

La población A tiene una tasa de crecimiento mayor porque se duplica más rápidamente.

# Población inicial

```python

N0 = 500
```

# Tiempo continuo

Estamos por definir una linea de tiempo de 20 horas, en las que vamos a pedir que se distribuyan en 200 tiempos uniformes. Entre mas tiempos más linda la gráfica. 

```python
tiempo = np.linspace(0, 20, 200)
```

### N(t)=N0 e^rt
 
donde:

##### N(t)	Población en el tiempo
##### N_0 	Población inicial 
##### e   	Número de Euler ≈ 2.718 
##### r	    Tasa de crecimiento
##### t	    Tiempo

# Modelo exponencial A

```python
modelo_A = N0 * np.exp(r_A * tiempo)
```

# Modelo exponencial B

```python
modelo_B = N0 * np.exp(r_B * tiempo)
```

```python
# Parámetros
N0 = 500
K = 100000
r = r_A

# Tiempo
tiempo = np.linspace(0, 20, 200)

# Modelo logístico
modelo_logistico = K / (1 + ((K - N0) / N0) * np.exp(-r * tiempo)
)

# Gráfica
plt.figure()

plt.plot(
    tiempo,
    modelo_logistico,
    color="black",
    label="Modelo logístico"
)

plt.axhline(
    K,
    linestyle="--",
    label="Capacidad de carga K"
)

plt.xlabel("Tiempo (horas)")
plt.ylabel("UFCS")
plt.title("Crecimiento logístico")

plt.legend()
plt.grid()

plt.show()
```

```python
# Parámetros
N0 = 500
K = 100000
r = r_A

# Tiempo
tiempo = np.linspace(0, 20, 200)

# Modelo exponencial
exponencial = N0 * np.exp(r * tiempo)

# Modelo logístico
logistico = K / (
    1 + ((K - N0) / N0) * np.exp(-r * tiempo)
)

# Gráfica
plt.figure(figsize=(10, 6))

plt.plot(
    tiempo,
    exponencial,
    color="red",
    label="Modelo exponencial"
)

plt.plot(
    tiempo,
    logistico,
    color="black",
    label="Modelo logístico"
)

plt.axhline(
    K,
    linestyle="--",
    label="Capacidad de carga K"
)

plt.xlabel("Tiempo (horas)")
plt.ylabel("UFCs")
plt.title("Crecimiento exponencial vs. crecimiento logístico")

plt.legend()
plt.grid()

plt.show()
```
